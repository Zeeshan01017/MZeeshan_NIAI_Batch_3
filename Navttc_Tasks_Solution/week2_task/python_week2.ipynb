```python
# Functions 
def get_marks(subjects):
    """Ask the user for marks in each subject and return them as a list."""
    marks = []
    for subject in subjects:
        while True:
            value = input(f"Enter marks for {subject} (0-100): ")
            if value.isdigit() and 0 <= int(value) <= 100:
                marks.append(int(value))
                break
            print("Invalid input. Enter a number between 0 and 100.")
    return marks


def calculate_total_and_average(marks):
    """Return total and average of the marks (two values)."""
    total = sum(marks)
    average = total / len(marks)
    return total, average


def get_grade(average):
    """Return a grade based on the average."""
    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 40:
        return "D"
    else:
        return "F"


def is_passed(marks, pass_mark=40):
    """Return True if the student passed every subject (pass_mark has a default value)."""
    for m in marks:
        if m < pass_mark:
            return False
    return True


def print_report(name, subjects, marks):
    """Print the final result card."""
    total, average = calculate_total_and_average(marks)
    grade = get_grade(average)
    status = "Pass" if is_passed(marks) else "Fail"

    print("\n----- Result Card -----")
    print(f"Name: {name}")
    for subject, mark in zip(subjects, marks):
        print(f"{subject}: {mark}")
    print(f"Total: {total}")
    print(f"Average: {average:.2f}")
    print(f"Grade: {grade}")
    print(f"Status: {status}")


def main():
    subjects = ["Math", "English", "Science"]
    name = input("Enter student name: ")
    marks = get_marks(subjects)
    print_report(name, subjects, marks)


main()
```

```python
# tuple is immutable, meaning it cannot be changed after creation. It is defined using parentheses ().
students = (("Ali", 80), ("Sara", 90), ("Ahmed", 70))

total = 0
for name, marks in students:
    print(f"{name}: {marks}")
    total += marks

print("Average:", total / len(students))

top = max(students, key=lambda s: s[1])
print("Topper:", top[0])
```

```python
# string datatype and set dtype
text = "Python is fun and Python is fun"

words = text.lower().split()       # convert to lowercase and split the string into a list of words
unique_words = set(words)          # a set removes duplicate words

print("Total words:", len(words))
print("Unique words:", len(unique_words))
print("Unique list:", sorted(unique_words))   # sets have no order, so sort before printing

vowels = set("aeiou")                          # set of all vowels
found = set(text.lower()) & vowels             # intersection: vowels that appear in the text

print("Vowels used:", sorted(found))
print("Missing vowels:", sorted(vowels - found))   # difference: vowels not in the text
```

```python
# Dictionary
contacts = {}                                 # empty dictionary: name -> phone

# Add contacts
contacts["Ali"] = "0300-1111111"
contacts["Sara"] = "0321-2222222"
contacts["Ahmed"] = "0333-3333333"

# Search for a contact
name = input("Search name: ").title()         # title() makes "ali" into "Ali"

if name in contacts:                          # check if the key exists
    print(f"{name}: {contacts[name]}")
else:
    print("Contact not found.")

# Show all contacts
print("\nAll contacts:")
for n, phone in contacts.items():             # loop over key and value together
    print(n, "->", phone)

print("Total contacts:", len(contacts))
```

```python
# Exception handling and user define exceptional handling
# User-defined exceptions (inherit from the built-in Exception class)
class InvalidAmountError(Exception):
    """Raised when the amount is zero or negative."""
    pass


class InsufficientBalanceError(Exception):
    """Raised when the balance is less than the withdrawal amount."""
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f"Balance Rs. {balance} is less than Rs. {amount}")


def withdraw(balance, amount):
    if amount <= 0:
        raise InvalidAmountError("Amount must be greater than zero.")   # raise our own exception
    if amount > balance:
        raise InsufficientBalanceError(balance, amount)
    return balance - amount


balance = 5000

try:
    amount = int(input("Enter amount to withdraw: "))    # may raise ValueError
    balance = withdraw(balance, amount)
    print(f"Success! Remaining balance: Rs. {balance}")

except ValueError:                                       # built-in exception
    print("Error: please enter numbers only.")

except InvalidAmountError as e:                          # user-defined exception
    print("Error:", e)

except InsufficientBalanceError as e:                    # user-defined exception
    print("Error:", e)
    print(f"You are short by Rs. {e.amount - e.balance}")

else:
    print("Transaction completed without errors.")      # runs only if no exception happened

finally:
    print("Thank you for using the ATM.")               # always runs
```
