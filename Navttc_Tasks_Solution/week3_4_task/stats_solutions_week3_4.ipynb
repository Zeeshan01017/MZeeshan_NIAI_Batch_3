```python
import statistics

data = [55, 60, 70, 70, 75, 80, 85, 90, 95, 100]
print("Data:", data)
# 1. MEAN 
mean = sum(data) / len(data)              # sum of all values / how many values
print("Mean:", mean)

# 2. MEDIAN 
sorted_data = sorted(data)                # median needs sorted data
mid = len(sorted_data) // 2               # middle position

if len(sorted_data) % 2 == 0:             # even count: average of the two middle values
    median = (sorted_data[mid - 1] + sorted_data[mid]) / 2
else:                                     # odd count: the single middle value
    median = sorted_data[mid]
print("Median:", median)

# 3. MODE 
mode = max(data, key=data.count)          # value that appears the most times
print("Mode:", mode)

# 4. STANDARD DEVIATION 
squares = [(x - mean) ** 2 for x in data]     # squared distance of each value from the mean
variance = sum(squares) / len(data)           # average of the squares
std = variance ** 0.5                         # square root of variance
print("Variance:", variance)
print("Std Deviation:", round(std, 2))

#  Check with the statistics library 
print("\nLibrary check:")
print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode(data))
print("Std (population):", round(statistics.pstdev(data), 2))
```
