```python
# Task_9: Python Program to solve a quadratic equation

import cmath

a = float(input("Enter the value of a: "))
b = float(input("Enter the value of b: "))
c = float(input("Enter the value of c: "))

discriminant = b**2 - 4*a*c

x1 = (-b + cmath.sqrt(discriminant)) / (2*a)
x2 = (-b - cmath.sqrt(discriminant)) / (2*a)

print(f"The roots are: {x1} and {x2}")
```

```python
# For Loop 
marks = [45, 82, 33, 67, 90]

pass_count = 0
total = 0

for i, m in enumerate(marks, start=1):
    if m >= 40:
        status = "Pass"
        pass_count += 1
    else:
        status = "Fail"
    total += m
    print(f"Student {i}: {m} -> {status}")

print("Total Pass:", pass_count)
print("Average Marks:", total / len(marks))
```

```python
# while loop 
total = 0
n = 1

while n <= 10:
    total += n
    n += 1

print("Sum:", total)
```

```python
# Looping techniques in python
import random

secret = random.randint(1, 20)
max_tries = 5
tries = 0
guesses = []

print("Number Guessing Game")
print("I have picked a number between 1 and 20.")

while tries < max_tries:
    user_input = input(f"Guess {tries + 1}/{max_tries}: ")

    # If the input is invalid, skip it (this does not count as a try)
    if not user_input.isdigit():
        print("Please enter numbers only!")
        continue

    guess = int(user_input)
    tries += 1
    guesses.append(guess)

    if guess == secret:
        print(f"Congratulations! You won in {tries} tries!")
        break
    elif guess < secret:
        print("The number is higher.")
    else:
        print("The number is lower.")
else:
    # Runs only when the loop finishes without hitting break
    print(f"You lost. The correct number was {secret}.")

print("\nYour guesses:")
for i, g in enumerate(guesses, start=1):
    print(f"{i}. {g}")
```
